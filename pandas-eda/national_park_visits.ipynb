# Exploring US National Park Visits
For this assignment, I decided to explore national park visits data. I wanted to see which national parks had the most visits in 2024 and how visitations between parks differed. 

```python
import pandas as pd
import altair as alt
```

```python
url = "https://raw.githubusercontent.com/melaniewalsh/responsible-datasets-in-context/main/datasets/national-parks/US-National-Parks_RecreationVisits_1979-2024.csv"

df = pd.read_csv(url)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
parks_2024 = df[df["Year"] == 2024]

top_parks = parks_2024.sort_values("RecreationVisits", ascending=False).head(10)

top_parks
```

```python
chart = alt.Chart(top_parks).mark_bar(color="pink").encode(
    x=alt.X("RecreationVisits:Q", title="Number of Visits"),
    y=alt.Y("ParkName:N",sort="-x", title="National Park"),
    tooltip=["ParkName", "State", "RecreationVisits"]
).properties(
    title = "Top 10 Most Visited National Parks in 2024"
)
chart
```

## Observations:

Something I found interesting about this dataset was how much the number of visits varied between national parks. I chose to focus on 2024 because I wanted to compare the data all within one year. The bar chart made is visually easy to compare visitation numbers instead of looking at each row of data. I also included tooltips so users can hover over each bar to see the exact number of visits. One limitation of this dataset is that it measures total visits instead of induvidual visitors which means it could count the same person more than once.
